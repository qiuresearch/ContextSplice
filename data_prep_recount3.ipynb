# Recount3 datasets (GTEx, SRA, and TCGA)

From: https://rna.recount.bio/docs/raw-files.html#project-level-count-files

For every project, we have files at the gene, exon, and exon-exon junction expression levels. For genes and exons, we provide a file for each of the annotations. That is, for every project we provide:

gene files: one count matrix per annotation
exon files: one count matrix per annotation
3 exon-exon junction files: the sparse count matrix data in Matrix Market (MM) format, the small list of sample identifiers (IDs), and the exon-exon junctions coordinate information (RR file)

```python
rnaseq_fastq_srr_biosample.sample(5)
# extract SRR ID from dbxrefs column
def extract_srr_id(dbxrefs):
    if pd.isna(dbxrefs):
        return None
    if ',' in dbxrefs:
        print("Multiple dbxrefs found:", dbxrefs, ' only using the first one.')
        dbxrefs = dbxrefs.split(',')[0]
    # srr_id = [ref.split(':')[1] for ref in dbxrefs.split(',') if ref.startswith('SRA:SRR')]
    if dbxrefs.startswith('SRA:SRR'):
        return dbxrefs.split(':')[1]
    else:
        return None

rnaseq_fastq_srr_biosample['SRR id'] = rnaseq_fastq_srr_biosample['dbxrefs'].apply(extract_srr_id)
# check how many non-empty SRR ids were extracted
num_srr_ids = rnaseq_fastq_srr_biosample['SRR id'].notna().sum()
print(f"Extracted {num_srr_ids} SRR IDs from dbxrefs (out of {len(rnaseq_fastq_srr_biosample)}).")
# check if SRR id matches the expression SRR[0-9]* pattern
import re
pattern = re.compile(r"SRR[0-9]+")
matching_srr_ids = rnaseq_fastq_srr_biosample['SRR id'].apply(lambda x: bool(pattern.fullmatch(x)) if pd.notna(x) else False).sum()
print(f"Number of SRR IDs matching pattern: {matching_srr_ids} out of {num_srr_ids} extracted.")

# remove rows without SRR id
# file_fastq_final = rnaseq_fastq_srr_biosample[rnaseq_fastq_srr_biosample['SRR id'].notna()].copy()
# print(f"Final RNA-seq fastq file report with SRR IDs shape: {file_fastq_final.shape}")
# file_fastq_final.sample(5)

encode_recount3_path = Path("data/encode/rnaseq_fastq_srr_biosample.pkl")
rnaseq_fastq_srr_biosample.to_pickle(encode_recount3_path)
rnaseq_fastq_srr_biosample.to_csv(encode_recount3_path.with_suffix('.csv'), index=False)
print(f"Saved final RNA-seq fastq file report with SRA and biosample metadata to {encode_recount3_path} and {encode_recount3_path.with_suffix('.csv')}")
```

```python
import requests
import pandas as pd

def query_from_biosample(biosample_id):
    """
    Given an ENCBS ID, find all related experiments and extract SRA IDs for RNA-seq.
    """
    # 1. Search for all experiments associated with this biosample
    search_url = f"https://www.encodeproject.org/search/?type=Experiment&replicates.library.biosample.accession={biosample_id}&format=json&limit=all"
    response = requests.get(search_url).json()
    
    results = []
    
    for exp in response.get('@graph', []):
        exp_id = exp.get('accession')
        assay = exp.get('assay_title')
        
        # Initialize record
        record = {
            "Biosample": biosample_id,
            "Experiment": exp_id,
            "Assay": assay,
            "SRRs": None
        }
        
        # 2. If it's RNA-seq, we need the SRR IDs for recount3
        if "RNA-seq" in assay:
            # We must fetch the full experiment JSON to see the file dbxrefs
            exp_data = requests.get(f"https://www.encodeproject.org/experiments/{exp_id}/?format=json").json()
            srr_list = []
            for file in exp_data.get('files', []):
                # Only look at raw Fastq files for SRA links
                if file.get('file_format') == 'fastq':
                    for ref in file.get('dbxrefs', []):
                        if 'SRA:SRR' in ref:
                            srr_list.append(ref.split(':')[1])
            record["SRRs"] = ",".join(list(set(srr_list)))
        
        results.append(record)
        
    return results

# Example Usage
my_biosamples = ["ENCBS046RNA", "ENCBS123ABC"] # Replace with your IDs
final_data = []
for bs in my_biosamples:
    final_data.extend(query_from_biosample(bs))

df = pd.DataFrame(final_data)
```
